```python
import requests
```

```python
BASE_URL = "http://localhost:8002/api/coffee"
```

```python
# 1. Probar Crear Finca
farm_data = {
    "name":"Finca El Salto",
    "zone":"Boquete",
    "origin":"Panamá"
}

response = requests.post(f"{BASE_URL}/farms",json=farm_data)
farm_id = response.json().get("id")
print(f"Status Finca: {response.status_code} | ID: {farm_id}")
```

```python
# 2. Probar Crear Lote Asociado
lot_data = {
    "variety":"Pacamara",
    "process":"Black Honey",
    "altitude":1750,
    "farm_id":farm_id
}
res_lot = requests.post(f"{BASE_URL}/lots", json=lot_data)
print(f"Status Lote: {res_lot.status_code} | Lote: {res_lot.json().get('variety')}")
```

```python
# 2. Probar obtener la Finca ID 1 con sus lotes
res_detail = requests.get("http://localhost:8002/api/coffee/farms/14")
data = res_detail.json()

if res_detail.status_code == 200:
    print(f"\nDetalle de Finca: {data['name']}")
    print(f"Lotes vinculados: {len(data['lots'])}")
    for lote in data['lots']:
        print(f"  - Variedad: {lote['variety']} | Proceso: {lote['process']}")
else:
    print("Error: Finca no encontrada.")
```
